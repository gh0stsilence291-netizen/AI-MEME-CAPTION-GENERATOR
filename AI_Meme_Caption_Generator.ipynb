# AI Meme Caption Generator

Lightweight + fun project. Situation describe karo, funny captions milenge (English + Hinglish).

**Kyun unique hai?**  
Creativity dikhata hai. Interview mein log isko yaad rakhte hain.

Pure Python — koi heavy model nahi.

```
import random
import re

TEMPLATES = [
    "When {desc}... me: {reaction}",
    "Nobody:\nAbsolutely nobody:\nMe after {desc}: {reaction}",
    "{desc}? Mere dimaag mein sirf {reaction}",
    "Expectation: {desc}\nReality: {reaction}",
    "Boss: {desc}\nMe: {reaction}",
    "That moment when {desc} and you just {reaction}",
    "POV: {desc}\nMy brain: {reaction}",
    "Me trying to {desc}: {reaction}",
    "Life after {desc}: {reaction}",
    "Everyone else: normal\nMe after {desc}: {reaction}",
]

REACTIONS = [
    "main toh gayab", "scene on hai", "kya bakchodi hai ye",
    "bhai ye kya ho raha hai", "full confusion", "brain 404 not found",
    "main kya karun ab", "silent treatment", "internal screaming",
    "just vibing in chaos", "this is fine", "send help",
    "why is life like this", "main nahi de raha exam", "system failure",
    "error 500", "bhai mat puch", "bas ho gaya aaj ka",
    "main so raha hu", "full drama mode"
]

HINGLISH_EXTRA = [
    "Arre bhai {desc} dekh ke mood off",
    "Ye {desc} dekh ke lagta hai zindagi cheat code use kar rahi hai",
    "Jab {desc} ho toh samajh lo universe tera mazak udha raha hai",
    "{desc} = instant depression button",
    "Bhai {desc} ke baad toh therapy chahiye",
]

def clean_desc(text):
    text = text.strip().lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    return text

def generate_captions(description, n=5):
    desc = clean_desc(description)
    if not desc:
        return ["Bhai kuch toh likh de situation ka"]
    
    captions = []
    for _ in range(n-1):
        template = random.choice(TEMPLATES)
        reaction = random.choice(REACTIONS)
        captions.append(template.format(desc=desc, reaction=reaction))
    
    captions.append(random.choice(HINGLISH_EXTRA).format(desc=desc))
    random.shuffle(captions)
    return captions

print("Meme engine ready ✅")
```

### Situation describe karo

```
desc = input("Situation / image describe karo (example: exam result aaya, salary day pe balance zero):\n")
print("\n🔥 Generated Captions:\n")
for i, cap in enumerate(generate_captions(desc, n=6), 1):
    print(f"{i}. {cap}")
    print()
```

```
# Aur generate karna ho to ye cell dubara chalao
desc2 = input("Naya situation daalo (ya same):\n")
print("\n🔥 New Captions:\n")
for i, cap in enumerate(generate_captions(desc2, n=5), 1):
    print(f"{i}. {cap}\n")
```
